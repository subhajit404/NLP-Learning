```python
import string
# translator = str.maketrans('', '', string.punctuation)
marks = string.punctuation
```

```python
def remove_punc(text):
    for char in marks:
        text= text.replace(char,'')
    return text
```

```python
df['text']=df['text'].apply(remove_punc)
```

```python
df
```

```python
import string
def remove_punc_ap(text):
    return text.translate(str.maketrans('','',string.punctuation))
```

```python
df['text']= df['text'].apply(remove_punc_ap)
```

```python
def reomve_number(txt):
    new=""
    for i in txt:
        if not i.isdigit():
            new+=i
    return new
```

```python
df['text']=df['text'].apply(reomve_number)
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('stopwords')
nltk.download('punkt')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
stop_words
```

```python
df.loc[1]['text']
```

```python
def remove_stop(txt):
    words = word_tokenize(txt)
    cleaned=[]
    for i in words:
        if not i in stop_words:
            cleaned.append(i)
    return  ' '.join(cleaned)
```

```python
df['text']= df['text'].apply(remove_stop)
```
