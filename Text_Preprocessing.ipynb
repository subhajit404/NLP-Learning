```python
import re
```

```python
def removal_html_tag(text):
    pattern =re.compile('<.*?>')
    return pattern.sub(r'',text)
```

```python
text = '<p>This is the second paragraph, separated by a line.</p>'

print(removal_html_tag(text))
```

```python
import string
# translator = str.maketrans('', '', string.punctuation)
marks = string.punctuation
```

```python
marks
```

```python
def remove_punc(text):
    for char in marks:
        text= text.replace(char,'')
    return text
```

```python
df['review']=df['review'].apply(remove_punc)
```

```python
def remove_punc1(text):
    return text.translate(str.maketrans('','',marks))
```

```python
df['review']=df['review'].apply(remove_punc1)
```

```python
import nltk

nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
stopwords.words('english')
```

```python
text ='I am going to the market because I need to buy some vegetables and fruits.'

def remove_stopwords(text):
    new_text =[]
    for words in text.split():
        if words in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(words)
    x= new_text[:]
    new_text.clear()
    return " ".join(x)

remove_stopwords(text)
```
